## 1. Project Objective:

This project explores taxi trip data to understand trip characteristics, fare patterns, temporal trends, geographic distribution, passenger behaviour, and data quality.

The analysis uses descriptive statistics, visualizations, aggregation, feature engineering, and hypothesis testing to identify patterns and relationships in the dataset.

A key objective is to distinguish genuine travel patterns from records that may distort analysis, while preserving the original data and documenting the limitations of the findings.

## 2. Key Finding
### Trip characteristics
- Trip distance and trip duration are right-skewed: most trips are relatively short, while a smaller number of trips have much larger values.
- Short trips (under 2 miles) account for approximately 53.8% of all records. Short and medium trips together account for approximately 81.3%.
- Typical trip values are better represented by medians than means when distributions contain extreme values.

### Temporal patterns
- The distribution of trip distances varies across pickup periods.
- In the filtered behavioural dataset, night trips have a lower share of short trips and a higher share of medium and long trips than other periods.
- These are observed associations; the analysis does not establish why the differences occur.

### Fare and payment patterns
- Weekday trips have a higher average fare than weekend trips in the analysed data, with a mean difference of approximately $1.33.
- The standardized effect size is small (Cohen's d ≈ 0.074), so statistical significance should not be interpreted as a large practical difference.
- Payment type and weekend status are statistically associated, but the association is small (Cramér's V ≈ 0.073).

### Data quality and engineered features
- Some records contain zero or unusually short durations, extreme distances, or unusually high derived rates.
- Ratios such as fare per minute and total per mile are sensitive to small denominators and extreme observations.
- Engineered flags and distance categories helped make these patterns easier to investigate without deleting the original records.

## 3. Data Quality and Limitations
- The dataset contains missing values in several columns, including passenger count and some fare-related or trip-classification fields.
- Zero-duration trips make duration-based rates undefined or unreliable.
- Extreme distance and duration records can strongly influence means, averages, and derived ratios.
- Statistical outliers are not automatically invalid records. Records were flagged for investigation rather than indiscriminately deleted.
- Some analyses use filtered dataframes with explicit conditions. Their results describe those subsets and should not be generalized automatically to the full dataset.
- The analysis is observational. Statistical associations do not establish causation.
- Findings describe the available dataset and may not represent all taxi trips, locations, or time periods.

## 4. Final Conclusion
This exploratory analysis identified important patterns in taxi trip distances, durations, fares, payment types, and pickup periods. Most recorded trips fall within the short and medium distance categories, while several distributions show substantial right-skewness.

The analysis also demonstrated how data quality issues and extreme observations can affect summary statistics and engineered features. Using diagnostic flags, filtered analysis dataframes, medians, and distribution percentiles helped provide a more careful interpretation of the data.

Overall, the project shows that effective EDA involves more than producing visualizations: it requires validating the data, selecting appropriate summaries, investigating unusual observations, and communicating the limitations of conclusions.

## 5. Recommendations for Further Analysis
- Investigate fare patterns across pickup and drop-off zones using clearly defined geographic groupings.
- Examine how fare components contribute to total trip amounts.
- Compare trip patterns across months or years, if the dataset covers a sufficiently broad time period.
- Build predictive models for trip duration or fare, after establishing a suitable validation strategy and handling data quality issues.
- Validate suspicious records against the original data source or documentation before treating them as errors.
